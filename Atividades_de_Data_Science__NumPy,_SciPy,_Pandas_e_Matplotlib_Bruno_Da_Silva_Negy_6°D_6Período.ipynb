```python
import numpy as np
from scipy.integrate import solve_ivp


def sistema(x, Y): #aqui eu separei em y que significa y e y_linha para derivados de y
    y = Y[0]
    y_linha = Y[1]

    dy_dx = y_linha
    dy_linha_dx = -y  #Pois y'' + y = 0 => y'' = -y

    return [dy_dx, dy_linha_dx]

solucao = solve_ivp(sistema, (0, 10), [1, 0]) #Resolve de x=0 a x=10, com y(0)=1 e y'(0)=0

print("Valores de y encontrados:", solucao.y[0])
```

```python
import matplotlib.pyplot as mp

listares =[] #listas vazias
listaint = []

def funcao(x): #funcao de x
  return 1/x

for i in range(1, 101): #for para 1 a 100
  listaint.append(i)# add no final da lista vazia
  listares.append(funcao(i))

mp.plot(listaint, listares) #as listas passadas sendo colocadas no grafico
mp.title('Y = 1/X com X de 1 a 100')
mp.xlabel('Valor de X:')
mp.ylabel('Valor de Y')
mp.show()
```

Criei as listas, depois disso fiz o calculo da funcao, defini x, e fiz o for para adicionar no fim da lista os itens, depois organizei o grafico

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100) #espaço

y = x ** 2 #exponencial de x

plt.plot(x, y)

plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.title('Função Quadrática')

plt.text(5, 25, 'x = 5') #texto de referencia

plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as mp

listares =[] #listas vazias
listaint = []

def funcao(x): #funcao de x
  return 1/x

for i in range(1, 101): #for com o limite
  listaint.append(i) #add nas listas
  listares.append(funcao(i))

mp.plot(listaint, listares) #criar o grafico
mp.title('Y = 1/X com X de 1 a 100')
mp.xlabel('Valor de X:')
mp.ylabel('Valor de Y')

mp.savefig("grafico.pdf") #salvar como pdf
```

#Exercicio 18

```python
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D', 'E', 'F'] #categorias
valores = [10, -5, 15, -8, 20, -12] #valores

cores = [] #vazio para cores

for valor in valores:
    if valor >= 0:
        cores.append('green') #adiciona cor se for maior
    else:
        cores.append('red') #se for menor coloca outra cor

plt.bar(categorias, valores, color=cores) #gera o grafico de barras com as cores

plt.axhline(0, color='black') #cor preto para linha

plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.title('Valores Positivos e Negativos')

plt.show()
```

Esse foi interessantew de fazer, nao utilizou conceitos muito complexos, acabeu somente gerando as categorias, e classificando com um for, e depois chamei o grafico com tudo
