#Exercicio 21: Busca Binaria

```python
lista = [3, 7, 12, 18, 25, 31, 40, 55] #Defini uma lista
numero = 25 #Numero definido no meio da lista

inicio = 0
fim = len(lista) - 1 #Começa no ultimo indice

while inicio <= fim: #Fiz um while pra definir a condição de inicio e fim
    meio = (inicio + fim) // 2 #Começa no meio

    if lista[meio] == numero: #If pra caso estiver no meio
        print("Encontrado na posição", meio) #Encontrado pois esta identificado no meio
        break

    elif numero < lista[meio]: #Else if pra caso for menor que o meio
        fim = meio - 1

    else:
        inicio = meio + 1 #Else pra caso for maior que o meio
else:
    print("Número não encontrado") #indicativo que nao foi encontrado, pois não esta no meio

```

#Exercicio 22: Criptografia de Cifra de César

```python
alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

texto = input("Digite uma palavra: ").upper() #Transforma as letras digitadas em maiuscula
deslocamento = int(input("Digite o quanto deseja deslocar: ")) #pede pra pessoa colocar o deslocamento

resultado = "" #Vazio pois ainda nao retornou criptografado

for letra in texto:
    if letra in alfabeto: #Verificar se esta no alfabeto digitado
        posicao = alfabeto.index(letra)  #Descobre em que posicao a letra esta
        nova_posicao = (posicao + deslocamento)% 26 #Gera a posição nova, e adiciona o deslocamento
        resultado += alfabeto[nova_posicao]
    else:
        resultado += letra

print("Texto criptografado e: ", resultado) #retorna o resultado tiver sucesso

```

#Exercicio 24: Validação de CPF

```python
def validar_cpf(cpf): #inicializa funcao
  if not cpf.isdigit(): #verifica se o cpf é composto de numeros
    return "CPF deve conter apenas numeros, CPF invalido"
  elif len(cpf) != 11: #verifica se o tamanho do cpf é correto
    return "Tamanho errado, CPF invalido"
  else: #de acordo com a regra oficial do cpf, os ultimos dois digitos são verificadores
    for i in range(9, 11):
        somas = sum(int(cpf[num]) * ((i + 1) - num) for num in range(0, i)) #multiplica cada um dos 9 iniciais por pesos que vãoo de 10 a 2
        digitos = ((somas * 10) % 11) % 10
        if digitos != int(cpf[i]):
            return "Erro ao validar, CPF invalido"
    return f"o cpf {cpf} é valido"#Sucesso

cpf = input("Digite o CPF")#pede e captura o cpf
print(validar_cpf(cpf))#mostra a funcao, passa o cpf e printa a mensagem
```

# Exercício 30: Jogo da Forca

```python

import random

lista_palavras = ["overwatch", "jogos", "widowmaker", "mercy", "valorant", "clove"] #Lista de palavras

palavra = random.choice(lista_palavras) #A palavra vai ser escolhida sendo opção de alguma da lista

acertos = ["_"] * len(palavra)
erros = []

chances = 5

print("=== JOGO DA FORCA ===")
print("Adivinhe a palavra!")
print(" ".join(acertos))
while chances > 0 and "_" in acertos: #While pra definir a condicao
    letra = input("Digite uma letra: ").lower()
    if len(letra) != 1 or not letra.isalpha(): #Verifica se so tem letras do alfabeto
        print("Digite somente uma letra.")#Verifica se foi digitada somente uma letra
        continue
    if letra in acertos or letra in erros:
        print("Você já tentou essa letra.")#Verifica se a letra já foi utilizada
        continue
    if letra in palavra:
        for posicao in range(len(palavra)):#Procura a letra dentro da palavra
            if palavra[posicao] == letra:
                acertos[posicao] = letra
        print("Você acertou uma letra!")
    else:
        erros.append(letra)
        chances -= 1
        print(f"Você errou! Restam {chances} chances.")
    print("Palavra:", " ".join(acertos))
    if erros:
        print("Letras erradas:", ", ".join(erros))
if "_" not in acertos:
    print(f"Parabéns! Você descobriu a palavra: {palavra}")#Mostra o resultado
else:
    print(f"Você perdeu! A palavra era: {palavra}") #Se perder

```

Achei esse exercicio bem complexo sendo sincero, ele é um conceito que conhecemos ja (jogo da forca), porém executar em pyhton não foi tao facil

#Exercicio 36:  Gerador de Senhas

```python
import random #Biblioteca random novamente

letras = "abcdefghijklmnopqrstuvwxyz" #alfabeto que pode ser utilizado
numeros = "0123456789" #numeros

tamanho = int(input("Digite o tamanho da senha: ")) #Input que pede o tamanho e converte a string digitada em int

senha = "" #Senha vazia pra começar

for i in range(tamanho): #Tamanho que o usuario desejar
  tipo = random.choice([letras, numeros]) #O tipo vai selecionar um dos 3
  senha += random.choice(tipo) #E aqui ele vai executar o random com o tipo aleatorio que cair

print("Sua senha é:", senha) #Mostra a senha
```


Fiz do jeito mais pratico usando a biblioteca random e definindo os tipos, foi bom

#Exercicio 37: Funções Recursivas

```python

def fatorial(numero): #Define a funcao

  if numero == 0 or numero == 1: #A quebra da recursão
    return 1

  return numero * fatorial(numero - 1) #Recursão

numero = int(input("Digite um número: ")) #Digitar o numero dom input

resultado = fatorial(numero) #Resultado

print("O fatorial de", numero, "é:", resultado)

```

#Exercucui 40: Lista de Compras

```python

lista_compras = [] #Criei a lista vazia pra iniciar

while True: #Repete até quebrar

  print("\n=== LISTA DE COMPRAS ===") #Opções
  print("1 - Adicionar item")
  print("2 - Remover item")
  print("3 - Mostrar lista")
  print("4 - Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == "1": #Se a opção for 1 ele adiciona

      item = input("Digite o item que deseja adicionar: ")

      lista_compras.append(item) #Adiciona na lista

      print("Item adicionado!")

  elif opcao == "2": #Se for 2 ele remove

      item = input("Digite o item que deseja remover: ")

      if item in lista_compras: #Verifica se o item esta na lista
          lista_compras.remove(item) #Remove
          print("Item removido!")
      else:
          print("Esse item não está na lista.")

  elif opcao == "3": #Se for 3 ele lista

      if len(lista_compras) == 0:
          print("A lista está vazia.") #Se estiver vazia ele mostra isso
      else:
          print("\nItens da lista:")

          for item in lista_compras: #Mostra os itens
              print("-", item)

  elif opcao == "4": #Se a opcao for 4 para sair

      print("Saindo...")
      break #O break

  else:
      print("Opção inválida.")
```

#Exercicio 42:  Números Primos até N

#Exercicio 45: Par ou Ímpar em Lista

```python

listan = [1, 3, 4, 10, 78, 100, 0] #Lista
listapar =[]
listaimpar = [] #Listas vazias
for n in listan: #Um for para passar pela lista contendo os numeros
  if n % 2 == 0: #se o resto da divisao for = a 0, indica que vai para lista par
    listapar.append(n)
  else: #caso o resultado da divisao nao de 0, ele vai para impar
    listaimpar.append(n)
print("Os números pares são: \n", listapar) #mostra as duas listas
print("Os numeros impares são: \n", listaimpar)
```

#Exercicio 46: Calcular Média e Desvio Padrão

```python

import statistics
lista = [10, 20, 30, 40, 60, 70, 80, 90, 100]

media = sum(lista) / len(lista) #calcular a media
desvio = statistics.stdev(lista) #formula de desvio padrao usando a biblioteca statistics

print("A média é: ", media)
print("O desvio padrão é: ", desvio)

```

#Exercicio 48: Ordenação por Inserção

```python

numeros = [5, 3, 8, 2, 7, 1] #Lista de Numeros

for i in range(1, len(numeros)): #Começa com o indice 1

    numero = numeros[i] #Iguala
    j = i - 1 #Faz o j mirar pro numero anterior

    while j >= 0 and numeros[j] > numero:
        numeros[j + 1] = numeros[j]
        j = j - 1

    numeros[j + 1] = numero #enquanto o numero for maior ele é empurrado pra direita

print("Lista ordenada:", numeros)

```

#Exercicio 50:  Cálculo de Frequência de Letras

```python

texto = input("Digite uma frase: ") #digitar uma frase
letras =[]
contagem = {}

for letra1 in texto: #passa por todas letras no texto digitado pela pessoa
  letras.append(letra1) #adiciona em uma lista vazia
for letra in letras:
    if letra in contagem: #se a letra estiver na contagem, o contador é incrementado
        contagem[letra] += 1
    else: #caso nao estiver, um novo é adicionado
        contagem[letra] = 1

print(contagem) #mostra a contagem

```
